```python
#16.4 use sqlite3 to create a sqlite database called book.dv and a table called books with firleds title, author and year
import sqlite3
conn = sqlite3.connect('books.db')
curs = conn.cursor()
curs.execute('''CREATE TABLE IF NOT EXISTS books
(title TEXT,
author TEXT,
year INTEGER
)''')
conn.commit()
curs.close()
conn.close()

print('books.db successfully created')
```

```python
#16.3 had to complete question 16.3 in order to complete 16.5
import csv
header = ['title', 'author', 'year']
data = [
['The Weirdstone of Brisingamen', 'Alan Garner', 1960],
['Perdido Street Station', 'China Miéville', 2000],
['Thud!', 'Terry Pratchett', 2005],
['The Spellman Files', 'Lisa Lutz', 2007],
['Small Gods', 'Terry Pratchett', 1992],
]
with open('books2.csv', 'w', newline='', encoding='utf-8') as file:
	writer = csv.writer(file)
	writer.writerow(header)
	writer.writerows(data)
print ('books2.csv created successfully')

#16.5 
import csv
import sqlite3
conn = sqlite3.connect('books.db')
curs = conn.cursor()
with open ('books2.csv', 'r', newline="", encoding='utf-8') as file:
	csv_reader = csv.DictReader(file)
	for row in csv_reader:
		curs.execute(" INSERT INTO books (title, author, year) VALUES(?,?,?)", (row['title'], row ['author'], int(row['year'])))
conn.commit()
curs.close()
conn.close()

print('Data sucessfully imported from books2.csv to books.db')
```

```python
#16.8 
import sqlalchemy as sa
conn = sa.create_engine('sqlite:///books.db')

#This version of sqlalchemy requires an explicit connection using context manager
with conn.connect() as connection:
	query = sa.text('SELECT * FROM books ORDER BY title ASC')
	rows = connection.execute(query)
for row in rows:
	print(row)
```
